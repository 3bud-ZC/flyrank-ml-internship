## 1. My lane (or freestyle) and why

I chose **Lane 2 — Refresh / Content Opportunity Scoring**. The goal is not to automate editorial decisions; it is to improve which pages a reviewer inspects first when review capacity is limited. The starter pipeline already demonstrates a transparent rule baseline, learned ranking, reason codes, and a review queue. I will use this lane provisionally and later strengthen the target with a future-window outcome from the warehouse if the data supports it.

```python
import os, subprocess
from pathlib import Path
import pandas as pd

REPO_URL = "https://github.com/3bud-ZC/flyrank-ml-internship"
REPO_DIR = "flyrank-ml-internship"

def find_repo_root():
    here = Path.cwd().resolve()
    for candidate in [here, *here.parents]:
        if (candidate / "data/raw/content_refresh_anonymized.csv").exists():
            return candidate
    return None

root = find_repo_root()
if root is None:
    if not Path(REPO_DIR).exists():
        subprocess.run(["git", "clone", "--depth", "1", REPO_URL, REPO_DIR], check=True)
    root = Path(REPO_DIR).resolve()

os.chdir(root)
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")
print("Working dir:", os.getcwd())
print("Rows:", len(df))
print("Clients:", df["client_id"].nunique())
print("Columns:", df.shape[1])
```

## 2. The question: decision, action, cost of a wrong call

**Research question:** Which content pages should be reviewed first for refresh or another content action, using observable search, engagement, freshness, and content signals while keeping the recommendation explainable and leakage-safe?

The decision is which pages enter a limited top-K editorial review queue first. A content strategist, SEO specialist, or editor acts on the output by reviewing the ranked candidates and choosing an action such as refresh, expand, review CTR/metadata, monitor, protect, merge, or prune. A false positive wastes limited editorial time and can trigger unnecessary changes to a healthy page; a false negative can leave a meaningful decline or opportunity unreviewed. The system is therefore decision support for human review, not an automatic publishing decision.

```python
review_capacity = 50
print("Illustrative review capacity (top-K):", review_capacity)
print("Evaluation focus: Precision@K, because the operational decision is which pages get reviewed first.")
```

## 3. Quick look at the data (2-3 real numbers)

The starter dataset contains **30,000 rows**. The starter proxy marks **16,262 rows as declining**, or about **54.2%**. The verified starter report also shows a meaningful ranking gap under client-holdout validation: the fixed baseline reaches **Precision@50 = 0.240**, while the random forest reaches **0.740**. These numbers make prioritization worth investigating further, while the final capstone should still use a stronger future-window target if the warehouse supports it.

```python
declining = df["trend_direction"].str.lower().eq("down")
visible = df["impressions_90d"] >= 500
stale = df["days_since_last_update"] >= 180

print("Total rows:", len(df))
print("Declining proxy rows:", int(declining.sum()))
print("Declining proxy rate:", round(declining.mean(), 3))
print("Visible pages (>=500 impressions):", int(visible.sum()))
print("Stale + visible pages:", int((visible & stale).sum()))
```
